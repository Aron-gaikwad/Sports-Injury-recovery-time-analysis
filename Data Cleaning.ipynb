```python
missing_cols = ['heart_rate', 'hydration_level', 'sleep_quality', 'muscle_activity', 
                'gait_speed', 'training_intensity']

# Apply median imputation
for col in missing_cols:
    if col in df.columns:
        df[col] = df[col].fillna(df[col].median())
```

```python
# Define medically and physiologically realistic boundaries
constraints = {
    "body_temperature": {"min": 34.0, "max": 42.0},  # Extreme hypothermia to severe hyperthermia
    "heart_rate": {"min": 35.0, "max": 220.0},       # Elite athletic resting heart rate to maximum exertion limit
    "hydration_level": {"min": 0.0, "max": 100.0}    # Percentage scale bounds
}

# Flag and handle anomalies
for column, bounds in constraints.items():
    if column in df.columns:
        # Count values breaking the rules
        invalid_count = df[(df[column] < bounds["min"]) | (df[column] > bounds["max"])].shape[0]
        print(f"Feature '{column}': Found {invalid_count} values outside realistic boundaries ({bounds['min']}°C - {bounds['max']}°C).")
        
        # Replace out-of-bounds anomalies with NaN so they can be accurately handled by MICE/Iterative Imputer
        df[column] = df[column].where((df[column] >= bounds["min"]) & (df[column] <= bounds["max"]), np.nan)

print("Logical consistency sweep complete. Anomalies flag-converted to NaN.")
```

```python
# Identify text-based categorical attributes
categorical_columns = ['playing_surface', 'sport_type', 'gender']

for col in categorical_columns:
    if col in df.columns:
        # Print unique items before cleaning to spot issues
        initial_uniques = df[col].dropna().unique().tolist()
        print(f"Original unique categories in '{col}': {initial_uniques}")
        
        # 1. Strip structural whitespace from the front and back of strings
        # 2. Force Title Case capitalization (e.g., 'track' or 'TRACK' -> 'Track')
        df[col] = df[col].astype(str).str.strip().str.title()
        
        # Treat any string converted to 'Nan' or 'None' back into true Numpy NaNs
        df[col] = df[col].replace({'Nan': np.nan, 'None': np.nan})
        
        # Print clean categories
        cleaned_uniques = df[col].dropna().unique().tolist()
        print(f"Cleaned unique categories in '{col}': {cleaned_uniques}\n")

print("String formatting and whitespace clipping complete.")
```

```python
# Cell 3: Dropping System and Database Identifiers
print("--- Cell 3: Feature Filtering ---")

# Specify purely administrative labels or relational keys
metadata_identifiers = ['athlete_id', 'session_id']

# Verify presence in your dataframe and drop them
existing_ids = [ident for ident in metadata_identifiers if ident in df.columns]

if existing_ids:
    print(f"Removing non-predictive structural columns from operational dataframe: {existing_ids}")
    # Drop columns inline or build an explicit modeling subset
    df_cleaned = df.drop(columns=existing_ids)
    print(f"Original shape: {df.shape} | New modeling subset shape: {df_cleaned.shape}")
else:
    df_cleaned = df.copy()
    print("No database identification indexes detected in dataset.")

print("Feature filtering complete. Dataframe is optimized for mathematical model evaluation.")
```
